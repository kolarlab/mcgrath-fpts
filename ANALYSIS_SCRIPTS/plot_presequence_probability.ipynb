```python
import json
from pathlib import Path

import matplotlib.pyplot as plt

BASE_DIR = Path("..")
PLOTS_DIR = BASE_DIR / "PLOTS"
PLOTS_DIR.mkdir(exist_ok=True)

PRESEQUENCE_PROBABILITY_THRESHOLD = 0.385

with open(BASE_DIR / "fpf_info_presequence_corrected.json") as f:
    fpf_info_corrected = json.load(f)

probabilities = []
for pdb_id, proteins in fpf_info_corrected.items():
    for protein in proteins:
        if "presequence_probability" in protein:
            probabilities.append(protein["presequence_probability"])

fig, ax = plt.subplots(figsize=(7, 5))
ax.hist(probabilities, bins=30, edgecolor="black")
ax.axvline(
    PRESEQUENCE_PROBABILITY_THRESHOLD,
    color="red",
    linestyle="--",
    linewidth=1.5,
    label=f"MitoFates threshold ({PRESEQUENCE_PROBABILITY_THRESHOLD})",
)
ax.set_xlabel("Predicted presequence probability")
ax.set_ylabel("Number of chains")
ax.set_title("MitoFates presequence probability distribution (mitochondrial chains)")
ax.legend()
fig.tight_layout()
fig.savefig(PLOTS_DIR / "presequence_probability_histogram.png", dpi=200)

print(f"Chains with a probability recorded: {len(probabilities)}")
print(f"Above threshold: {sum(1 for p in probabilities if p >= PRESEQUENCE_PROBABILITY_THRESHOLD)}")
print(f"Below threshold: {sum(1 for p in probabilities if p < PRESEQUENCE_PROBABILITY_THRESHOLD)}")
print(f"Plot written to {PLOTS_DIR}/presequence_probability_histogram.png")
```
